```python
url = 'https://feeds.datagolf.com/get-player-list?file_format=[ file_format ]&key=8c29c8197a3391551e4092d3dd9b'
data = requests.get(url).json()

amateurBool = []
country = []
countryCode = []
dg_id = []
name = []

for player in data:
    amateurBool.append(player['amateur'])
    country.append(player['country'])
    countryCode.append(player['country_code'])
    dg_id.append(player['dg_id'])
    name.append(player['player_name'])

# Create a DataFrame from the lists
df = pd.DataFrame({
    'amateur': amateurBool,
    'country': country,
    'country_code': countryCode,
    'dg_id': dg_id,
    'name': name
})

df.head()

df.to_csv('datagolf_player_list.csv', index=False)
```

Field Updates

Data Golf Rankings

```python
# Returns top 500 players in current DG rankings, along with skill estimate and OWGR rank

url = 'https://feeds.datagolf.com/preds/get-dg-rankings?file_format=[ file_format ]&key=8c29c8197a3391551e4092d3dd9b'
data = requests.get(url).json()

dg_rank = []; dg_ID = []; dg_skill_estimate = []; owgr_rank = []; primary_tour = []

last_update = data['last_updated']
rankings = data['rankings']

for player in rankings:
    dg_rank.append(player['datagolf_rank'])
    dg_ID.append(player['dg_id'])
    dg_skill_estimate.append(player['dg_skill_estimate'])
    owgr_rank.append(player['owgr_rank'])
    primary_tour.append(player['primary_tour'])

# Create a DataFrame from the lists
df = pd.DataFrame({
    'datagolf_rank': dg_rank,
    'dg_id': dg_ID,
    'dg_skill_estimate': dg_skill_estimate,
    'owgr_rank': owgr_rank,
    'primary_tour': primary_tour
})
df['last_updated'] = last_update

df.head()
df.to_csv('datagolf_rankings.csv', index=False)
```

Player Skill Estimates

```python
url = 'https://feeds.datagolf.com/preds/skill-ratings?display=[ display ]&file_format=[ file_format ]&key=8c29c8197a3391551e4092d3dd9b'
data = requests.get(url).json()

dg_id = []; driving_acc=[]; driving_dist = []; sg_app = []; sg_arg = []; sg_ott = []; sg_putt = []; sg_total = []

last_update = data['last_updated']
ratings = data['players']

for player in ratings:
    dg_id.append(player['dg_id'])
    driving_acc.append(player['driving_acc'])
    driving_dist.append(player['driving_dist'])
    sg_app.append(player['sg_app'])
    sg_arg.append(player['sg_arg'])
    sg_ott.append(player['sg_ott'])
    sg_putt.append(player['sg_putt'])
    sg_total.append(player['sg_total'])

# Create a DataFrame from the lists
df = pd.DataFrame({
    'dg_id': dg_id,
    'driving_accuracy': driving_acc,
    'driving_distance': driving_dist,
    'sg_approach': sg_app,
    'sg_around_green': sg_arg,
    'sg_off_the_tee': sg_ott,
    'sg_putting': sg_putt,
    'sg_total': sg_total
})
df['last_updated'] = last_update

df.head()
df.to_csv('player_skill_ratings.csv', index=False)
```

Historical Raw Event IDs

```python
# get list of tournaments and corresponding IDs for history, use this to get detailed info on each tournament
# note - event IDs are identical for same event from year to year, need to use calendar year to differentiate, also tour since majors show up on multiple tours
url = 'https://feeds.datagolf.com/historical-raw-data/event-list?file_format=[ file_format ]&key=8c29c8197a3391551e4092d3dd9b'
data = requests.get(url).json()

calendar_year = []; event_id = []; event_name = []; sg_categories = []; tour = []; traditional_stats = []

for event in data:
    calendar_year.append(event['calendar_year'])
    event_id.append(event['event_id'])
    event_name.append(event['event_name'])
    sg_categories.append(event['sg_categories'])
    tour.append(event['tour'])
    traditional_stats.append(event['traditional_stats'])

# Create a DataFrame from the lists
df = pd.DataFrame({
    'calendar_year': calendar_year,
    'event_id': event_id,
    'event_name': event_name,
    'sg_categories': sg_categories,
    'tour': tour,
    'traditional_stats': traditional_stats
})

df.head()

df.to_csv('historical_tournaments.csv', index=False)
```
