```python
import pandas as pd
import numpy as np
max_rows = 1000000
max_cols = 1000
pd.set_option('display.max_rows', max_rows)
pd.set_option('display.max_columns', max_cols)
```

```python
player_data = pd.read_csv('nhl_players.csv')
```

```python
player_data = player_data.drop([0])
```

```python
player_data.head()
```

```python
player_data['isActive'] = player_data['isActive'].astype(str).str.lower().map({'true': True, 'false': False})
```

```python
player_data['isActive'].value_counts()
```

```python
active_players = player_data[player_data['isActive']==True]
```

```python
active_players.fillna(0, inplace=True)
```

```python
active_players['regSeasonCareerGoals'] = active_players['regSeasonCareerGoals'].astype(int)
```

```python
active_players.sort_values('regSeasonCareerGoals', ascending=False, inplace=True)
```
